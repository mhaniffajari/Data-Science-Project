## `JOIN` Statement

Statement `JOIN` digunakan untuk **menggabungkan dua tabel** menjadi satu tabel melalui **kolom penghubung** yang sama. Operasi `JOIN` dapat dilakukan lebih dari satu kali dalam satu SQL query. Berikut adalah diagram ilustrasi beberapa macam operasi SQL `JOIN`:

![](assets/sqljoins.png)

💡 `JOIN` yang sering digunakan:
- `LEFT JOIN` paling sering digunakan pada sebagian besar skenario bisnis. `LEFT JOIN` mengembalikan semua baris pada tabel kiri **terlepas** dari apakah ada baris yang cocok pada tabel kanan.
- `INNER JOIN` adalah tipe join yang sangat intuitif dan mudah dipahami. Query ini mengembalikan semua baris di tabel kiri yang cocok dengan tabel kanan.

**Note**:

- `RIGHT JOIN` hampir tidak pernah digunakan karena ekuivalen dengan `LEFT JOIN` hanya mengganti peletakkan tabel kiri dan kanannya saja.
- `FULL JOIN` sangat jarang digunakan. Selain alasan komputasi, tipe join ini mengembalikan semua baris dari kedua tabel terlepas dari apakah ada kecocokan atau tidak, sehingga menghasilkan DataFrame dengan banyak missing value.

Ilustrasi terkait ke-empat tipe join dapat dilihat pada [Visual JOIN](https://joins.spathon.com/).

Syntax `JOIN`:

```sql
SELECT <COLUMNS>
FROM <LEFT_TABLE>

[LEFT|INNER|FULL] JOIN <RIGHT_TABLE>
ON <LEFT_TABLE>.key = <RIGHT_TABLE>.key
```

Perhatikan kembali skema database `chinook.db`:

1. Tabel `albums`: 
    - `AlbumId`
    - `Title`
    - `ArtistId`

 2. Tabel `artists`:
    - `ArtistId`
    - `Name` 
    
❓**Kasus:** Kita ingin mendapatkan DataFrame yang menampilkan semua judul album (`Title`) dari tabel `albums` secara lengkap, ditambahkan dengan informasi beserta kolom `Name`. Perhatikan bahwa kolom `Name` terdapat pada tabel `artists`, sedangkan kolom `AlbumId` dan `Title` terdapat pada tabel `albums`.

- `LEFT JOIN`
- Tabel kiri: albums
- Tabel kanan: artists
- Penghubung: ArtistId

📝 **Note**: Ketika menuliskan nama-nama kolom pada query yang menggunakan 2 tabel atau lebih harus menambahkan nama tabelnya sebelum menuliskan naam kolomnya dengan format: `table.column`

````python
# your code here
query = 'SELECT albums.Title, artists.Name FROM albums LEFT JOIN artists ON albums.ArtistId = artists.ArtistId'

pd.read_sql_query(query, conn)
````

📌 **Tips:**

String pada python yang kita gunakan (`'query'`) hanya dapat menyimpan teks/query dalam satu line code. Jika kita gunakan untuk menuliskan query yang panjang maka akan cukup sulit untuk pemahamannya. 

Sebagai alternatif, anda dapat mengapit SQL query dengan `'''query'''` (petik tiga kali) agar dapat menuliskan multiline query:

````python
# code here
query = '''
    SELECT albums.Title, artists.Name 
    FROM albums 
    LEFT JOIN artists 
    ON albums.ArtistId = artists.ArtistId
'''
pd.read_sql_query(query, conn)
````

Terkadang penamaan kolom ataupun tabel pada database cukup panjang dan redundan untuk diketik, ataupun belum cukup informatif.

📌 **Tips:** <br>
Penggunaan statement `AS` dapat digunakan untuk melakukan **aliasing / rename** nama tabel maupun nama kolom.

````python
# code here
query = '''
    SELECT al.Title AS AlbumTitle, ar.Name AS ArtistName
    FROM albums AS al
    LEFT JOIN artists AS ar
    ON al.ArtistId = ar.ArtistId
'''
pd.read_sql_query(query, conn)
````

⚠️ **Hati-hati!**<br>
Saat melakukan `LEFT JOIN`, peletakkan tabel kiri dan kanan sangat berpengaruh terhadap hasil penggabungan.

❓Sebagai perbanding, sekarang coba kita tukar tabel `artists` menjadi tabel kiri, sedangkan `albums` menjadi tabel kanan.

Object `df` di bawah menampilkan semua data `artists` walaupun artist tersebut tidak ada di tabel `albums`.

````python
# FROM artists LEFT JOIN albums
query = '''
    SELECT albums.Title AS AlbumTitle, artists.Name AS ArtistName
    FROM artists
    LEFT JOIN albums
    ON albums.artistId = artists.artistId
'''

df = pd.read_sql_query(query, conn)
df
````

## `GROUP BY` Statement

Di `pandas`, kita bisa menggunakan metode `.crosstab()` untuk membuat tabel agregasi. Sedangkan di SQL, kita menggunakan kombinasi statement `GROUP BY` dan menggunakan beberapa fungsi agregasi pada statement `SELECT`:

Berikut beberapa fungsi agregasi yang tersedia pada SQL:

```sql
- COUNT()
- SUM()
- AVG()
- MIN()
- MAX()
```

❓ Pada tabel `invoices`, misalkan kita ingin mengetahui jumlah besar transaksi (`Total`) dan berapa transaksi yang dilakukan untuk setiap `CustomerId`:

````python
# your code here
query = '''
    SELECT CustomerId, SUM(Total) AS TotalSales, COUNT(CustomerId) AS Frequency
    FROM invoices

    GROUP BY CustomerId
'''

pd.read_sql_query(query, conn).head()
````

💡 Kemudian, jika misalkan kita ingin mengetahui CustomerId top 5 yang memiliki transaksi terbesar, maka kita dapat gunakan `ORDER BY` dan menentukan apakah kita ingin mengurutkannya dengan:

- `ASC`: mengurutkan dari kecil ke besar
- `DESC`: mengurutkan dari besar ke kecil

````python
# your code here
query = '''
    SELECT CustomerId, SUM(Total) AS TotalSales, COUNT(CustomerId) AS Frequency
    FROM invoices

    GROUP BY CustomerId
    ORDER BY TotalSales DESC
'''

pd.read_sql_query(query, conn)
````

## `WHERE` Statement

Sekarang, kita akan melihat teknik untuk melakukan **conditional subsetting** atau **filter baris** menggunakan statement `WHERE` yang diikuti dengan **kondisi**.

### Comparison Operator

Statement `WHERE` dapat dikombinasikan dengan comparison operator seperti: 

- Mathematical Operator : `=`, `!=`, `>`, `>=`, `<`, dan `<=`
- Logical Operator : `AND`, `OR`, dan `NOT`:

Sebagai contoh, misalkan kita ingin melakukan analisis terhadap semua data `invoices` yang ada di `BillingCountry` Germany, maka kita bisa menambahkan statement `WHERE` sebagai berikut:

````python
# your code here
query = '''
    SELECT *
    FROM invoices

    WHERE BillingCountry = 'Germany'
'''

pd.read_sql_query(query, conn)
````

### `IN` Operator

Operator `IN` memungkinkan kita menentukan beberapa nilai untuk perbandingan. Misalnya, seperti pada kasus sebelumnya, kita ingin mengambil semua data `invoices` yang terjadi di negara (`BillingCountry`) Canada dan juga USA:

````python
# your code here
query = '''
    SELECT *
    FROM invoices

    WHERE BillingCountry IN ('Canada', 'USA')
'''

pd.read_sql_query(query, conn)
````

### `LIKE` Operator

Operator `LIKE` sangat berguna jika kita perlu mencocokan bagian tertentu dari sebuah string daripada menggunakan operator sama dengan (`=`).

Sebagai contoh, kita ingin mendapatkan informasi customer yang tinggal pada daerah yang `BillingPostalCode` **dimulai** dengan angka 107. 

Kita dapat menggunakan `%` yang disebut sebagai [wildcard character](https://www.w3schools.com/sql/sql_wildcards.asp)

````python
# your code here
query = '''
    SELECT *
    FROM invoices

    WHERE BillingPostalCode LIKE '107%'
'''

pd.read_sql_query(query, conn)
````

❓Bagaimana jika anda menggunakan wildcard `%` diawal string?

````python
# your code here
query = '''
    SELECT *
    FROM invoices

    WHERE BillingPostalCode LIKE '%01'
'''

pd.read_sql_query(query, conn)
````

**Diskusi:**

- Jika `BillingPostalCode LIKE '107%'`, maka semua baris dengan kode pos **dimulai** angka 107 akan tampil.
- Jika `BillingPostalCode LIKE '%107'`, maka semua baris dengan kode pos **diakhiri** angka 107 akan tampil.

❓ Menurut Anda apa yang akan muncul jika kita menggunakan `%` **sebelum dan sesudah** pencocokan pola?

````python
# your code here
query = '''
    SELECT *
    FROM invoices

    WHERE BillingPostalCode LIKE '%107%'
'''

pd.read_sql_query(query, conn)
````

## `HAVING` Statement

Kita telah mempelajari bagaimana melakukan agregasi di SQL dan penggunaan pernyataan `WHERE`. Namun, terdapat sebuah masalah ketika kita ingin menggunakan `WHERE` untuk memfilter baris berdasarkan kondisi pada kolom yang sudah diagregasi.  

Oleh karena itu kita perlu menggunakan statement lain yaitu `HAVING`. Perbedaan utama antara `HAVING` dan `WHERE` adalah bahwa klausa `HAVING` digunakan setelah fungsi agregasi. 

Perhatikan query dibawah ini di mana kita ingin mengambil data pelanggan dengan total jumlah invoice sepanjang waktu lebih besar dari 47:


````python
# code here
query = """
    SELECT CustomerId, SUM(Total) as TotalValue
    FROM INVOICES

    GROUP BY CustomerId
    HAVING TotalValue > 47
"""

pd.read_sql_query(query, conn)
````

## SQL Subquery

Dalam beberapa kasus tertentu, kita ingin melakukan filter baris berdasarkan syarat tertentu dimana nilai-nilai kondisi hanya dapat dihitung menggunakan hasil dari query lainnya. Umumnya, subquery dapat didefinisikan dengan menuliskan query didalam query yang dibungkus dengan tanda kurung:

```sql
SELECT column1
FROM tables1
WHERE (
    SELECT column2
    FROM tables2
)
```

Pada section ini, kita akan membahas beberapa jenis subqueries yang dapat kita gunakan:

### Scalar Subquery

**Scalar subquery** mengembalikan satu nilai saja (satu baris dan satu kolom). Subquery jenis ini sering digunakan ketika hanya satu nilai yang diharapkan, misalnya dalam perhitungan dan perbandingan pada statement `WHERE`.

Sebagai contoh, kita ingin menampilkan informasi `CustomerId` dan `Total` dari tabel `invoices` yang memiliki `Total` diatas rata-rata:

````python
# code here
query = """
    SELECT AVG(Total)
    FROM invoices
"""

pd.read_sql_query(query, conn)
````

````python
# code here
query = """
    SELECT CustomerId, Total
    FROM invoices
    
    WHERE Total > (SELECT AVG(Total) FROM invoices)
"""

pd.read_sql_query(query, conn)
````

### Single-Column Subquery

**Single-Column subquery** memberikan informasi dari satu kolom dan beberapa baris. Subquery jenis ini sering digunakan ketika hanya terdapat beberapa nilai dalam suatu kolom dibutuhkan, misalnya ketika menggunakan statement `IN` pada pada `WHERE`.

Sebagai contoh, kita ingin menampilkan informasi lagu-lagu (`Name`) dari tabel `tracks` yang dibeli oleh Top 5 `CustomerId` dengan `Total` dari tabel `invoices` terbesar.

Pertama-tama, kita akan mempersiapkan informasi untuk memperoleh informasi Top 5 Customer tersebut:

````python
# code here
query = """
    SELECT CustomerId
    FROM invoices

    GROUP BY CustomerId
    ORDER BY SUM(Total) DESC

    LIMIT 5
"""

pd.read_sql_query(query, conn)
````

Sebenarnya, Untuk menarik informasi yang dibutuhkan, kita bisa saja menuliskan list `CustomerId` dalam kondisi secara **hard-code** seperti ini:

````python
query = """
    SELECT invoices.CustomerId, tracks.Name
    FROM invoices

    LEFT JOIN invoice_items ON invoice_items.InvoiceId = invoices.InvoiceId
    LEFT JOIN tracks ON tracks.TrackId = invoice_items.TrackId

    WHERE CustomerId IN (6, 26, 57, 45, 46)
"""

pd.read_sql_query(query, conn)
````

Namun cara tersebut tidak disarankan. Jika ada penambahan value dalam database, sehingga bisa saja bukan lagi `CustomerId` 6, 26, 57, 45, 46 yang merupakan top 5 customers. Maka dari itu, kita perlu menggunakan **subquery** agar query kita lebih **dinamis terhadap perubahan data**.

Untuk itu mari kita persiapkan subquery yang hanya mengembalikan list Top 5 `CustomerId` berdasarkan `Total` pembelian:

````python
# code here
query = """
    SELECT invoices.CustomerId, tracks.Name
    FROM invoices

    LEFT JOIN invoice_items ON invoice_items.InvoiceId = invoices.InvoiceId
    LEFT JOIN tracks ON tracks.TrackId = invoice_items.TrackId

    WHERE CustomerId IN (
        SELECT CustomerId
        FROM invoices

        GROUP BY CustomerId
        ORDER BY SUM(Total) DESC

        LIMIT 5
    )
"""

pd.read_sql_query(query, conn)
````

Kedua tabel akan menghasilkan hasil yang sama, namun query yang digunakan akan lebih dinamis.

### Multi-Column Subquery

**Multi-Column subquery** memberikan informasi dari beberapa kolom sekaligus. Subquery jenis ini sering digunakan jika kita butuh mengambil beberapa kriteria dari beberapa kolom, misalnya ketika menggunakan statement `IN` pada pada `WHERE`.

Sebagai contoh, dari tabel `invoices`, kita ingin menampilkan informasi transaksi `InvoiceId`, `InvoiceDate` manakah yang memiliki `Total` terbesar untuk masing-masing `CustomerId`:

Pertama-tama, kita mencoba untuk mempersiapkan informasi untuk memperoleh nilai `Total` maksimum untuk tiap customer:

````python
# code here
query = """
    SELECT CustomerId, MAX(Total)
    FROM invoices

    GROUP BY CustomerId
"""

pd.read_sql_query(query, conn)
````

Kemudian, informasi diatas dapat dijadikan kriteria yang akan difilter dengan menggunakan `WHERE`:

````python
# code here
query = """
    SELECT CustomerId, InvoiceId, InvoiceDate, Total
    FROM invoices

    WHERE (CustomerId, Total) IN (
        SELECT CustomerId, MAX(Total)
        FROM invoices

        GROUP BY CustomerId    
    )
"""

pd.read_sql_query(query, conn)
````

````python
# SUB QUERY pada statement SELECT 
query = """
    SELECT *, (SELECT AVG(Total) FROM invoices)
    FROM invoices
"""

pd.read_sql_query(query, conn)
````

````python
# SUB QUERY pada statement FROM 
query = """
    SELECT *
    FROM (        
        SELECT CustomerId
        FROM invoices

        GROUP BY CustomerId
        ORDER BY SUM(Total) DESC

        LIMIT 5
    ) AS top5customer

    LEFT JOIN invoices ON invoices.CustomerId = top5customer.CustomerId
    
"""

pd.read_sql_query(query, conn)
````

````python
# CTE
query = """
    WITH top5customer AS (
        SELECT CustomerId
        FROM invoices

        GROUP BY CustomerId
        ORDER BY SUM(Total) DESC

        LIMIT 5
    )

    SELECT *
    FROM top5customer

    LEFT JOIN invoices ON invoices.CustomerId = top5customer.CustomerId
    
"""

pd.read_sql_query(query, conn)
````

## Temporary Tables

Temporary tables adalah sebuah tabel untuk menyimpan hasil sementara selama menjalankan query yang kompleks tanpa mengubah tabel utama dalam database. Temporary tables hanya akan ada selama sesi SQL/Python kita berjalan, berfungsi sebagai workspace sementara untuk transformasi dan agregasi data.  

Temporary tables sangat berguna saat menangani operasi multi-langkah, seperti join beberapa tabel dan melakukan agregasi yang kompleks. Sehingga, temporary tables akan membantu memecah query kompleks menjadi bagian-bagian yang lebih kecil dan mudah dikelola.  

Untuk membuat temporary table, syntax umumnya adalah:  

```sql
CREATE TEMPORARY TABLE temp_table_name AS
```

Untuk pembuatannya di python, tidak menggunakan `pd.read_sql_query()` karena pembuatan temporary table tidak akan menampilkan informasi tabel tersebut, namun menggunakan `conn.execute()`

Kita akan melakukan query top 5 customer yang sebelumnya kita lakukan menggunakan subquery, namun sekarang kita akan buat menggunakan temporary tables:

````python
# code here
query = '''
    CREATE TEMPORARY TABLE top5customer AS
        SELECT CustomerId
        FROM invoices

        GROUP BY CustomerId
        ORDER BY SUM(Total) DESC

        LIMIT 5        
'''

conn.execute(query)
````

Setelah mendefinisikan temporary table bernama `top5_customer`, kita dapat menggunakannya dalam query kita pada bagian `FROM`:

````python
# code here
query = """
    SELECT *
    FROM top5customer
"""

pd.read_sql_query(query, conn)
````

Kemudian kita akan lengkapi query tersebut untuk melihat lagu-lagu yang dibeli customer tersebut:

````python
query = """
    SELECT invoices.CustomerId, tracks.Name
    FROM top5customer

    LEFT JOIN invoices ON invoices.CustomerId = top5customer.CustomerId
    LEFT JOIN invoice_items ON invoice_items.InvoiceId = invoices.InvoiceId
    LEFT JOIN tracks ON tracks.TrackId = invoice_items.TrackId
"""

pd.read_sql_query(query, conn)
````

### 🤿 Dive Deeper: Temporary Tables

Sebelumnya, kita juga sudah melakukan pembuatan subquery untuk melihat `invoices` transaksi terbesar (`Total`) untuk masing-masing `CustomerId` sehingga kita dapat melihat `InvoiceId` dan `InvoiceDate`nya.

```sql
SELECT CustomerId, InvoiceId, InvoiceDate, Total
FROM invoices

WHERE (invoices.CustomerId, invoices.Total) IN (

    FROM invoices
    GROUP BY CustomerId
)
```

Gunakan temporary table untuk menghasilkan output yang sama dengan query tersebut!

💡 **Hint**: JOIN dapat menggunakan 2 atau lebih kriteria dengan menggunakan `AND`/`OR`

````python
# code here - data retrieval
query = """
    SELECT invoices.CustomerId, InvoiceId, InvoiceDate, Total
    FROM max_customer

    LEFT JOIN invoices
    ON invoices.CustomerId = max_customer.CustomerId AND invoices.Total = max_customer.MaxTotal
"""

pd.read_sql_query(query, conn)
````

## Window Function

**Window Functions** adalah salah satu fitur paling powerful di SQL yang memungkinkan kita melakukan perhitungan pada sekumpulan baris sambil tetap menjaga setiap baris individual dalam hasil. Berbeda dengan fungsi aggregate seperti `GROUP BY` yang menggabungkan baris menjadi satu output, window functions mengembalikan nilai untuk setiap row, memberikan konteks sekaligus mempertahankan detail.

Window functions didefinisikan berpasangan dengan statement `OVER()`, di mana kita bisa menentukan:
- `PARTITION BY` untuk membagi data ke dalam kelompok.
- `ORDER BY` untuk menentukan urutan di dalam setiap kelompok.

Struktur umumnya adalah:

```sql
SELECT column_name1, 
       WINDOW_FUNCTION(column_name2) OVER (PARTITION BY column_name3 ORDER BY column_name4) AS new_column
FROM table_name
```

- `WINDOW_FUNCTION`: Window functions yang akan diterapkan
- `column_name1`: Kolom biasa yang akan ditampilkan di output
- `column_name2`: Kolom yang digunakan pada window function
- `column_name3`: Kolom yang digunakan untuk membagi rows ke dalam grup (`PARTITION BY`)
- `column_name4`: Kolom yang digunakan untuk menentukan urutan rows dalam setiap grup (`ORDER BY`)

Pada bagian ini, kita akan mengeksplorasi berbagai jenis window functions.

### Aggregate Window Function

**Aggregate Window Functions** mengubah sifat dari aggregate functions biasa dengan menerapkannya pada setiap row tanpa menggabungkan hasil menjadi satu record, seperti cara kerja `GROUP BY`. Maka, kita bisa menghitung statistik pada tiap kategori namun tetap mempertahankan setiap baris di output.

Untuk Aggregate Window Functions, biasanya kita menggunakan aggregate functions yang sudah dikenal di SQL, tetapi digabungkan dengan klausa `OVER()`. Berikut adalah yang paling umum digunakan:

- `SUM(column)`   : cumulative totals, running sums  
- `AVG(column)`   : moving averages, average per group  
- `COUNT(column)` : jumlah rows dalam sebuah window  
- `MIN(column)`   : nilai terkecil dalam window  
- `MAX(column)`   : nilai terbesar dalam window  

Sebagai contoh, kita mungkin ingin menghitung `Total` untuk setiap `CustomerId` serta menampilkan seluruh informasi kolom-kolom lainnya dari tabel `invoices`:

````python
query = '''
    SELECT *, SUM(Total) AS TotalPerCustomer
    FROM invoices
    GROUP BY CustomerId
'''

pd.read_sql_query(query, conn).head(20)
````

Dapat kita lihat bahwa proses tersebut hanya akan menampilkan 1 baris untuk masing-masing customer dan hanya akan menampilkan informasi pertama saja untuk kolom-kolom lainnya. 

Jika kita ingin menampilkan seluruh informasi lainnya setelah melakukan agregasi, kita dapat menggunakan metode window functions seperti berikut:

````python
# code here
query = '''
    SELECT *, SUM(Total) OVER(PARTITION BY CustomerId) AS TotalPerCustomer
    FROM invoices
'''

pd.read_sql_query(query, conn).head(20)
````

📝 Tambahkan `ORDER BY InvoiceDate` pada window function untuk melakukan akumulasi jumlah transaksi tiap waktu!

````python
# code here - running sum/cummulative total
query = '''
    SELECT *, SUM(Total) OVER(PARTITION BY CustomerId ORDER BY InvoiceDate) AS TotalPerCustomer
    FROM invoices
'''

pd.read_sql_query(query, conn).head(20)
````

### Ranking Window Functions

**Ranking Window Functions** digunakan untuk memberikan urutan atau ranking untuk tiap kategori yang ditentukan. Berbeda dengan aggregate window functions yang meringkas nilai, ranking functions membantu mengidentifikasi posisi relatif dari setiap row berdasarkan aturan sorting dari sebuah kategori.

Berikut adalah beberapa Window Functions yang paling umum digunakan untuk ranking:

- `ROW_NUMBER()` : Memberikan nomor urut unik secara berurutan pada setiap row dalam result set (tidak ada duplikasi, tidak ada celah).  
- `RANK()` : Memberikan ranking dimulai dari 1, tetapi akan meninggalkan gap dalam ranking jika ada nilai yang sama.  
- `DENSE_RANK()` : Mirip dengan `RANK()`, tetapi tidak meninggalkan gap jika ada nilai yang sama.  
- `NTILE(n)` : Membagi rows menjadi `n` kelompok yang kurang lebih sama besar (tiles atau buckets), dan memberikan setiap row ke salah satu kelompok tersebut.  

Kemudian dilanjutkan oleh statement `OVER()`, untuk mengatur:

- `PARTITION BY` untuk menentukan kolom kategori setiap urutan.
- `ORDER BY` untuk menentukan urutan berdasarkan nilai kolom apa.

Sebagai contoh, berikan ranking besarnya `Total` tiap `CustomerId` dari tabel `invoices`!

````python
# code here
query = '''
    SELECT *, RANK() OVER(PARTITION BY CustomerId ORDER BY Total DESC) AS RankingTotalPerCustomer
    FROM invoices
'''

pd.read_sql_query(query, conn).head(20)
````

Bandingkan dengan jenis-jenis ranking window functions lainnya!

DENSE RANK = Akan menghilangkan gap di antara ranking yang nilainya sama dengan nilai setelahnya

````python
# code here
query = '''
    SELECT *, DENSE_RANK() OVER(PARTITION BY CustomerId ORDER BY Total DESC) AS RankingTotalPerCustomer
    FROM invoices
'''

pd.read_sql_query(query, conn).head(20)
````

````python
# code here
query = '''
    SELECT *, ROW_NUMBER() OVER(PARTITION BY CustomerId ORDER BY Total DESC) AS RankingTotalPerCustomer
    FROM invoices
'''

pd.read_sql_query(query, conn).head(20)
````

````python
# code here
query = '''
    SELECT *, NTILE(4) OVER(PARTITION BY CustomerId ORDER BY Total DESC) AS RankingTotalPerCustomer
    FROM invoices
'''

pd.read_sql_query(query, conn).head(20)
````

### Value Window Functions

**Value Window Functions** memungkinkan KITA untuk mereferensikan nilai dari baris lain dalam sebuah window data yang sudah didefinisikan. Berbeda dengan agregasi atau ranking, fungsi ini melihat nilai dari sekitar baris tersebut untuk mengambil informasi. Sehingga, fungsi ini sangat berguna dalam analisis time-series, pelacakan transaksi customer dari waktu-ke-waktu, dan perbandingan row-to-row di mana konteks dari baris yang berdekatan menjadi penting.

Contoh paling umum dari **Value Window Functions** adalah:

- `LAG(column, offset)` : mengembalikan nilai dari baris sebelumnya dalam partition yang sama.
- `LEAD(column, offset)` : mengembalikan nilai dari baris setelahnya dalam partition yang sama.
- `FIRST_VALUE(column)` : mengembalikan nilai pertama dalam ordered window.
- `LAST_VALUE(column)` : mengembalikan nilai terakhir dalam ordered window.

Misalnya kita ingin membandingkan setiap transaksi customer dengan transaksi sebelumnya:


````python
# code here
query = '''
    SELECT *, LAG(Total, 1) OVER(PARTITION BY CustomerId ORDER BY InvoiceDate) AS PreviousTransaction
    FROM invoices
'''

pd.read_sql_query(query, conn).head(20)
````

Bandingkan dengan jenis-jenis ranking window functions lainnya!

````python
# code here
query = '''
    SELECT *, LEAD(Total, 2) OVER(PARTITION BY CustomerId ORDER BY InvoiceDate) AS PreviousTransaction
    FROM invoices
'''

pd.read_sql_query(query, conn).head(20)
````

````python
# code here
query = '''
    SELECT *, FIRST_VALUE(Total) OVER(PARTITION BY CustomerId ORDER BY InvoiceDate) AS PreviousTransaction
    FROM invoices
'''

pd.read_sql_query(query, conn).head(20)
````

````python
# code here
query = '''
    SELECT *, LAST_VALUE(Total) OVER(PARTITION BY CustomerId) AS PreviousTransaction
    FROM invoices
'''

pd.read_sql_query(query, conn).head(20)
````

# Under and Over Fetching

Di antara semua tools yang telah kita pelajari untuk menganalisis data, sekarang saatnya kita merenungkan yang mana yang lebih cocok untuk Anda. Untuk meninjau, mari kita ingat kembali apa yang telah kita pelajari:

- Reading flat files (CSV file)
- Data cleansing and wrangling
- Exploratory data analysis tools
- Visual exploratory tools

Kira-kira SQL masuk ke bagian apa? Untuk menjawab hal tersebut, Anda perlu memahami arsitektur client-server.

![](assets/clientserver.png)

Ketika Anda bekerja dengan SQL, kemungkinan besar Anda memiliki database relasional yang disimpan terpusat pada _server_ dan dapat diakses oleh _client_. Pada dasarnya _server_ adalah komputer penyedia data, dan _client_ adalah komputer pengambil data (komputer Anda).

Saat Anda melakukan query, sebenarnya Anda sedang menjalankan perintah untuk mengunduh data ke komputer Anda (_local computer_). Proses pengunduhan ini membutuhkan sumber daya, sehingga pengunduhan perlu dilakukan secara efektif untuk meminimalkan biaya.

- **Over** fetching adalah kondisi ketika kita menarik data dari database **lebih dari** yang dibutuhkan, sehingga membutuhkan biaya dan waktu yang lebih.

- **Under** fetching adalah kondisi sebaliknya, dimana data yang ditarik **kurang dari** yang dibutuhkan, sehingga proses analisis tidak dapat dilakukan secara lengkap.
