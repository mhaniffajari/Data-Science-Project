### Concept: Linear Regression



Regresi Linier adalah metode statistik yang digunakan untuk memodelkan hubungan antara dua variabel dengan mencocokkan garis lurus pada data. Metode yang paling umum untuk menentukan "garis paling sesuai" ini Ordinary Least Squares (OLS). 

#### Ordinary Least Squares (OLS)

OLS berfungsi untuk menemukan garis yang meminimalkan jumlah kuadrat jarak vertikal dari setiap titik data ke garis tersebut.



Perhatikan contoh regresi linier menggunakan OLS diatas, kedua gambar menunjukkan data observasi (titik biru), garis regresi (garis merah), dan jarak dari setiap titik ke garis. Jarak vertikal inilah yang kita sebut sebagai residual atau error, yaitu selisih antara nilai aktual (y) dan nilai yang diprediksi oleh model.

​
Gambar kanan mengilustrasikan inti dari metode OLS: alih-alih meminimalkan residual itu sendiri, OLS bertujuan untuk meminimalkan jumlah kuadrat setiap residual yang direpresentasikan oleh besar persegi merah. Dengan demikian, tujuan algoritma OLS adalah untuk menemukan kemiringan (slope) dan titik potong (intercept) dari garis merah yang membuat total luas dari semua kotak merah tersebut menjadi sekecil mungkin, sehingga menghasilkan garis yang paling pas dengan sebaran data kita.

#### OLS Formula

Hasil dari OLS dapat didefinisikan dengan persamaan berikut:

$\hat{y} = a + b*x $

dimana:
- $\hat{y}$ = Nilai prediksi
- $x$ = Variabel prediktor
- $a$ = Intercept, titik potong garis ke sumbu $y$, nilai $y$ ketika $x$ adalah 0
- $b$ = Slope, kemiringan garis, nilai koefisien dari $x$


> Formula dari OLS secara visual: https://setosa.io/ev/ordinary-least-squares-regression/

### Machine Learning Modelling

Berdasarkan eksplorasi yang sudah kita lakukan sebelumnya, kita akan mencoba untuk membuat sebuah model regresi linier untuk memprediksi besar minyak bumi yang diproduksi (`BORE_OIL_VOL`) berdasarkan tekanan pada wellhead (`AVG_WHP_P`) yang digunakan.

Untuk membuat model regresi linier, kita akan menggunakan fungsi-fungsi yang disediakan oleh library `statsmodels`. Jalankan kode berikut untuk mengimport library tersebut:

````python
import statsmodels.api as sm
````

#### <u>**Tahapan 1 - Menentukan Target dan Prediktor**</u>

Secara default, `statsmodels` tidak akan otomatis menambahkan intercept ke dalam model, maka model yang dibangun akan dipaksa melewati titik nol (0,0). 

Oleh karena itu, untuk menambahkan Intercept dalam model, kita perlu menambahkan sebuah nilai konstan pada data kita sebagai sebuah kolom menggunakan fungsi `sm.add_constant()`


   - `x` = `sm.add_constant(df['prediktor'])`.
   - `y` = `df['target']`


````python
# menambahkan intercept/add_constant ke prediktor
x = sm.add_constant(volve['AVG_WHP_P'])

# membuat variabel target
y = volve['BORE_OIL_VOL']
````

````python
x
````

#### <u> **Tahapan 2 - Fitting Model Regresi Linier**</u>

Untuk membuat model regresi linier di Python kita akan menggunakan fungsi `OLS(target, prediktor)` dari package `statsmodels` dan kemudian dilanjutkan dengan method `.fit()` untuk melatih model pada data kita.

````python
# Inisialisasi model OLS (regresi linier)
ols = sm.OLS(y, x)

# Fitting model OLS
model_ols = ols.fit()
````

#### <u>**Tahapan 3 - Melihat Hasil / Menginterpretasikan Model** </u>

Setelah model regresi linier dibangun, langkah berikutnya adalah menginterpretasikan output model. Dengan melakukan interpretasi ini, kita dapat menyimpulkan faktor mana yang benar-benar penting untuk memprediksi target serta menilai apakah model sudah cukup baik untuk digunakan dalam pengambilan keputusan.

Kita dapat memperoleh ringkasan hasil dari model secara lengkap dengan menggunakan fungsi `.summary()`

````python
# Summary model
model_ols.summary()
````

Output dari `.summary()` model OLS menampilkan banyak sekali informasi yang dapat kita manfaatkan untuk menginterpretasi model, namun sebenarnya tidak semua informasinya relevan terhadap tujuan akhir kita. Oleh karena itu, kita akan coba breakdown dan membahas nilai-nilai apa saja yang perlu diperhatikan:

##### Coefficients (`.params`)

Nilai koefisien (`coef`) merupakan nilai yang digunakan oleh OLS untuk menentukan Intercept (`const`) dan Slope (`AVG_WHP_P`) dari model. 

`BORE_OIL_VOL` $=$ `const` $+$  `coef` $*$ `AVG_WHP_P`

Maka, dari model tersebut, kita dapat menggunakan `.params` untuk melihat coefficient dari kedua variabel tersebut:

````python
# Model Coefficients
model_ols.params
````

Dapat kita simpulkan bahwa persamaannya menjadi:

`BORE_OIL_VOL` $= -884.6785 + 46.5180  * $ `AVG_WHP_P`

> Interpretasi:

   - Intercept : Jika `AVG_WHP_P` tidak diatur bernilai 0, maka `BORE_OIL_VOL` akan bernilai $-884.6785$ barrel.
   - Slope    : setiap kenaikan 1 nilai dari `AVG_WHP_P`, akan meningkatkan `BORE_OIL_VOL` sebesar $46.5180$ barrel.


##### P-Values (`.pvalues`)

**P-value** (`P>|t|`) adalah nilai hasil dari uji hipotesis, dimana dapat membantu kita untuk menentukan apakah variabel prediktor benar-benar punya pengaruh yang signifikan terhadap target. 

* $H_0$: Koefisien regresi bernilai 0  
    - Jika **p-value ≥ 0.05** → Variabel prediktor tidak berpengaruh terhadap target.
* $H_1$: Koefisien regresi berbeda secara signifikan dari nilai 0
    - Jika **p-value < 0.05** → Variabel prediktor punya pengaruh signifikan terhadap target.

````python
# P Values
model_ols.pvalues
````

````python
if model_ols.pvalues[1] < 0.05:
    print(f"Model is significant with p-value: {model_ols.pvalues[1]:.4f}")
else:
    print("Model is not significant.")
````

### Leverage and Influence

Dalam analisis regresi linier, tidak semua data memiliki pengaruh yang sama terhadap hasil model. **Leverage** adalah nilai yang letaknya jauh dari letak observasi lainnya, sering disebut sebagai **outlier**. Nilai leverage dapat mempengaruhi (**Influence**) model linier regresi atau pun tidak bergantung dengan posisinya:

- Ketika **leverage mempengaruhi (menurunkan R-Squared)** model linier regresi: **high influence**, sebaiknya **di-exclude** -> membuat model menjadi lebih buruk
- Ketika **leverage tidak mempengaruhi (meningkatkan R-Squared)** model linier regresi: **low influence**, sebaiknya **di-include** -> membuat model menjadi lebih baik

### Normality of Residual

**Harapannya ketika membuat model linear regression**, error yang dihasilkan berdistribusi normal. Artinya error banyak berkumpul disekitar angka 0. Untuk mengecek residual menyebar normal, pengujian yang paling sering dilakukan adalah Shapiro test:
- $H_0$: Residual berdistribusi normal
- $H_1$: Residual tidak berdistribusi normal

Dalam melakukan pengujiannya kita akan dibantu library `scipy` dan memanfaatkan fungsi `shapiro()`. Untuk memanfaatkan fungsi tersebut, kita akan mengeluarkan nilai residu dari model yang sudah dibuat dengan menambahkan `.resid` pada objek model yang dibuat.

📌 **Note**: Jika asumsi normalitas tidak terpenuhi, maka hasil uji signifikansi serta nilai standard error dari intercept dan slope setiap prediktor yang dihasilkan bersifat bias atau tidak mencerminkan nilai sebenarnya. Jika residual memiliki distribusi yang tidak normal, bisa lakukan **transformasi/scaling data pada target variabel** atau **menambahkan sample data**.

````python
from scipy.stats import norm

x = np.linspace(start=np.min(model_multiple.resid), stop=np.max(model_multiple.resid), num=100)   # buat 100 titik antara min-max
p = norm.pdf(x, loc=np.mean(model_multiple.resid), scale=np.std(model_multiple.resid))  # hitung nilai PDF normal di titik x

plt.plot(x, p, 'r', linewidth=2, label="Distribusi Normal");

plt.hist(model_multiple.resid, bins=50, density=True);
````

Selain itu bisa juga menggunakan pengujian statistik yaitu **Shapiro Test**. Dalam melakukan pengujiannya kita akan dibantu library `scipy` dan memanfaatkan fungsi `shapiro()`. Untuk memanfaatkan fungsi tersebut, kita akan mengeluarkan nilai residu dari model yang sudah dibuat dengan menambahkan `.resid` pada objek model yang dibuat.

````python
from scipy.stats import shapiro

name = {0: 'Statistics', 1: 'p-value'}
pd.DataFrame({shapiro(model_multiple.resid)}).rename(columns=name)
````

### Homoscedasticity

Homocesdasticity menunjukkan bahwa residual atau error bersifat konstan atau tidak membentuk pola tertentu. Jika error membentuk pola tertentu seperti garis linear atau mengerucut, maka kita sebut dengan `Heterocesdasticity` dan akan berpengaruh pada nilai standard error pada estimate/koefisien prediktor yan bias (terlalu sempit atau terlalu lebar).

Untuk mengecek terjadinya heteroscedasticity kita dapat menggunakan Breusch-Pagan test:
- $H_0$: residual homogen(tidak membentuk sebuah pola/acak)
- $H_1$: residual heteros (membentuk sebuah pola)

Kita bisa menggunakan method `het_breuschpagan()` dari library `statsmodels`.

### No Multicolinearity

- y - x -> korelasinya kuat
- x - x -> korelasinya harus kecil

Harapannya pada model linear regression, tidak terjadi multikolinearitas. Multikolinearitas terjadi ketika antar variabel prediktor yang digunakan pada model memiliki hubungan yang kuat. Ada atau tidak multikolinearitas dapat dilihat dari nilai VIF(Variance Inflation Factor). 

VIF dibagi menjadi beberapa nilai berikut:
- 1 = tidak berkorelasi antar prediktornya
- antara 1 dan 5 = korelasinya moderate
- Lebih besar 5 = paling kuat berkorelasi antar prediktornya
- Biasanya VIF lebih besar 10 adalah yang menunjukkan variabel prediktor sangat berkorelasi kuat.

> Ketika nilai VIF > 10 maka **ada hubungan yang kuat antar prediktor**. Yang diingkan ketika membuat model, nilai VIF < 10 agar **tidak ada hubungan antar prediktor**.

````python
from statsmodels.stats.outliers_influence import variance_inflation_factor as vif
import statsmodels.api as sm

vif = [vif(x_train_multi, i) for i in range(len(x_train_multi.columns))]
pd.Series(data=vif, index = x_train_multi.columns).sort_values(ascending=False)
````

### Why Assumption Matters?

Jika salah satu/lebih uji asumsi tidak terpenuhi, model sebenarnya masih bisa dijalankan dan dimanfaatkan. Namun, terdapat beberapa interpretasi dari model yang tidak akurat atau bahkan menyesatkan bergantung dengan asumsi manakah yang tidak terpenuhi:

1. **Linearitas**

    - Coefficient regresi salah merepresentasikan hubungan antar variabel.
    - Dampak: Model tidak bisa menangkap pola sebenarnya dan error prediksi akan tinggi.

2. **Normalitas residual**

    - Coefficient regresi tidak terpengaruhi
    - Nilai P-value bisa menyesatkan (koefisien terlihat signifikan padahal tidak, atau sebaliknya).
    - Dampak: Salah interpretasi dalam signifikansi variabel.

3. **Homoscedasticity**

    - Coefficient regresi tidak terpengaruhi
    - Nilai P-value bisa menyesatkan (koefisien terlihat signifikan padahal tidak, atau sebaliknya).
    - Dampak: Salah interpretasi dalam signifikansi variabel.

4. **No Multicolinearity**

    - Coefficient regresi tidak stabil, bisa memiliki nilai yang sangat besar.
    - Dampak: Model tetap bisa prediksi dengan baik, namun interpretasi tiap coefficient variabel tidak bisa dipercaya.

## Model Evaluation

Sebelumnya kita sudah membahas R-squared, yang mengukur seberapa baik model regresi dapat merepresentasikan data. Namun, perlu diperhatikan bahwa nilai R-squared yang dihasilkan dari model biasanya dihitung hanya pada data training, yaitu data yang digunakan untuk melatih model.

Untuk mendapatkan gambaran yang lebih objektif, kita perlu melakukan evaluasi menggunakan data test yang tidak pernah dilihat saat pelatihan. Sehingga, kita perlu memanfaatkan beberapa nilai berikut untuk melihat seberapa baik performa model kita pada data baru:

1. **MAE (Mean Absolute Error)**: Mengukur rata-rata selisih absolut antara nilai aktual dan prediksi. **Formula:**
   $$
   MAE = \frac{1}{N} \sum_{i=1}^{N} \left | y_{i} - \hat{y_i} \right |
   $$

2. **MAPE (Mean Absolute Percentage Error)**: Mengukur rata-rata persentase error relatif terhadap nilai aktual. (📝**Note**: Hanya dapat digunakan jika tidak ada nilai 0 pada data aktual). **Formula:**
   $$
   MAPE = \frac{1}{N} \sum_{i=1}^{N} \frac {\left | y_{i} - \hat{y_i} \right |} {y_i}
   $$

3. **RMSE (Root Mean Squared Error)**: Mengukur rata-rata kuadrat error. Ketika nilai error besar, maka nilai RMSE akan semakin besar dan sebaliknya. **Formula:**
   $$
   RMSE = \sqrt{\frac{1}{N} \sum_{i=1}^{N} (y_{i} - \hat{y_i})^{2}}
   $$
     
RMSE digunakan ketika model yang dibuat memuat observasi outlier, MAE digunakan ketika model yang dibuat tidak memuat observasi outlier, dan MAPE adalah metrik yang baik untuk interpretasi karena mudah dipahami.

### Model Improvement: Tuning Hyperparameter SVR

**Hyperparameter** adalah parameter-parameter yang dapat diatur oleh pembuat model untuk mempengaruhi cara belajar dari sebuah model. Dengan mengatur *hyperparameter* yang tepat, kita dapat meningkatkan performa model kita.

Analogi sederhana dari mengatur *hyperparameter* ini adalah adalah seperti instruksi seorang koki ketika membuat resep, ada beberapa hal yang dapat diatur yang dapat mempengaruhi cita rasa makanan, seperti jumlah garam, waktu memasak, atau suhu oven. Hal-hal tersebut adalah *hyperparameter* dalam membuat sebuah hidangan.

Adapun *hyperparameter* yang sering digunakan dalam model SVR antara lain:

- `C` : Regularisasi mengontrol margin dan jumlah kesalahan klasifikasi, semakin besar nilai C maka semakin kecil jumlah kesalahan, semakin kecil nilai C, semakin besar nilai kesalahan yang ditoleransi.
	* Nilai default = `1`
	* Range Input = `0.0 - inf`
	* Semakin besar nilai semakin kompleks

- `kernel` : Mempengaruhi bagaimana bentuk dari *hyperplane*, untuk data yang dapat dipisahkan secara garis linear (garis lurus) dapat menggunakan `linear`. Untuk data-data non-linear dapat menggunakan kernel `rbf`.
	* Nilai default = `'rbf'`
	* Opsi Input = `'rbf', 'poly', 'linear', 'sigmoid'`
    * Note: Kebanyakan data berbentuk non-linear, sehingga sebaiknya selalu memilih kernel `rbf`

Untuk hyperparameter dan penjelasan lebih detail dapat dilihat pada [dokumentasi SVR](https://scikit-learn.org/stable/modules/generated/sklearn.svm.SVR.html#sklearn.svm.SVR)

Mari kita mengatur hyperparameter dari model kita untuk mendapatkan performa yang lebih baik.

````python
# tuning dan melakukan pelatihan
model_svr_tuned = model_svr.set_params(C=100000, kernel='rbf')

model_svr_tuned.fit(x_train_scale, y_train_multi)
````

# Additional: Other Machine Learning Model

Selain regresi linier, terdapat banyak algoritma lain dalam machine learning yang dapat digunakan untuk membangun model prediksi. Algoritma ini seringkali lebih fleksibel dalam menangkap pola yang kompleks pada data, terutama ketika hubungan antara variabel tidak bersifat linear atau ketika jumlah fitur sangat banyak.

1. **Decision Tree**

    Decision Tree adalah salah satu algoritma machine learning yang populer karena konsepnya sederhana namun cukup kuat untuk memprediksi baik data kategorik maupun numerik.     Keunggulan utama Decision Tree adalah kemudahannya untuk dipahami dan dijelaskan, karena alur pengambilan keputusan dapat divisualisasikan layaknya diagram pohon. Selain itu, Decision Tree mampu menangkap hubungan non-linear antar variabel tanpa memerlukan uji asumsi klasik seperti pada regresi linier.

````python
from sklearn.tree import DecisionTreeRegressor

model_dt = DecisionTreeRegressor()
model_dt.fit(x_train_multi, y_train_multi)
````

2. **Random Forest**

    Random Forest adalah algoritma machine learning berbasis ensemble yang dibangun dari kumpulan banyak Decision Tree. Berbeda dengan pohon tunggal yang rentan terhadap overfitting, Random Forest menggabungkan hasil prediksi dari banyak pohon untuk menghasilkan prediksi yang lebih stabil dan akurat. Setiap pohon dalam Random Forest dilatih menggunakan subset acak dari data (bootstrap sampling) dan subset acak dari fitur (feature bagging), sehingga masing-masing pohon belajar pola yang berbeda. Pada tahap prediksi, hasil dari semua pohon digabungkan, dengan cara rata-rata untuk regresi.

    Keunggulan Random Forest antara lain memiliki performa yang lebih baik dibanding Decision Tree tunggal, lebih tahan terhadap outlier, serta mampu memberikan informasi penting mengenai variabel (feature importance).
