## Cross-Validation

Sebelum kita membuat model, kita harus membagi dataset menjadi data train dan test. Silahkan bagi data dengan proporsi 80% train dan 20% test.

Dalam melakukan splitting data silahkan gunakan `random_state = 0`.

```python
# splitting data
from sklearn.model_selection import train_test_split
# your code here
# define predictor & target variable
y = soil['soil_type']
x = soil.drop(columns='soil_type')

# cross validation
x_train, x_test, y_train, y_test = train_test_split(x, y, 
                                                      random_state=0, 
                                                      test_size=0.2)

```

# Model Fitting

## Decision Tree

Mari buat model Decision Tree dan simpan ke dalam objek `model_dt`. Untuk melakukan *tuning* model, mari kita atur parameter :

- `max_depth = 7`
- `min_samples_leaf = 20`
- `min_sample_split = 10`
- `random_state=123`

```python
# your code here
from pyexpat import model
from sklearn.tree import DecisionTreeClassifier

# create model here
model_dt = DecisionTreeClassifier(random_state=123,max_depth=7,min_samples_leaf=20,min_samples_split=10)
model_dt.fit(x_train, y_train)
```

```python
# visualization
from sklearn.tree import plot_tree
plt.figure(figsize=(15,15), dpi=150)
plot_tree(model_dt, 
          feature_names=x_train.columns, 
          filled=True, 
          rounded=True);
```

    - [ ] Jumlah minimum observasi pada internal node adalah 10. Jika tidak terpenuhi, maka tidak dilakukan percabangan.
    - [ ] Jumlah minimum observasi pada simpul daun adalah 10
    - [ ] Jumlah simpul internal yang terbentuk pada pohon adalah 10

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree, export_text
plt.figure(figsize=(20,10), dpi=300)
plot_tree(model_dt,
          feature_names=x_train.columns,
          max_depth=2,
          class_names=['clays', 'gravelly sands to dense sand', 'organic soil',
       'sand mixtures', 'sands', 'silt mixtures'],
          filled=True,
          rounded=True);
```

## Random Forest

Model terakhir yang ingin kita buat adalah Randon Forest. Di bawah ini merupakan keunggulan dari model Random Forest:

- Automatic feature selection
- Mereduksi bias pada model karena melakukan aggregasi dari hasil beberapa decision tree
- Menghasilkan estimasi out-of-box error yang tidak bias

Sekarang, coba buat model random forest dan simpan model dalam objek `model_rf`. Model tersebut dibuat menggunakan *hyperparameter* di bawah ini:

- `n_estimator = 250` 
- `random_state = 123` 
- `ood_score = True`



```python
# fill in the blank

from sklearn.ensemble import RandomForestClassifier

# create and fitting your model here
model_rf = RandomForestClassifier(n_estimators=250,random_state=123,oob_score=True)

```

Dalam praktiknya, random forest telah memiliki estimasi out-of-bag (OOB) yang merepresentasikan akurasi pada *out-of-bag data* (data yang tidak diambil ketika sampling/tidak digunakan dalam pembuatan random forest).

# Evaluasi Model 

Buatlah confusion matrix pada model decisio  tree maupun random forest lalu carilah nilai accuracy, recall, dan precision untuk masing-masing model.


```python
# fill in the blank code
from sklearn.metrics import accuracy_score, recall_score, precision_score, classification_report

# model Decision Tree
print(classification_report(
    y_true=y_test,
    y_pred=y_pred_dt,
    target_names=['clays', 'gravelly sands to dense sand', 'organic soil',
       'sand mixtures', 'sands', 'silt mixtures'],
    digits=5))
```
