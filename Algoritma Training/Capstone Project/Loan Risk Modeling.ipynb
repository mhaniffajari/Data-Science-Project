**Problem**

A financial company has **8%** creditors who default. This issue can lead to a loss of **$138,46,851,949**. The company wants the Data Science Team to collaborate with financial analysts to solve this problem. The company aims to assign the right interest rates and credit limits to minimize losses.

**Business Metrics**

1. Minimize losses caused by defaulting creditors
2. Reduce the number of creditors who default

**Goals**

1. Build a model that can perform credit scoring for applicants as a basis for determining the appropriate interest rates and credit limits

```python
def preprocessing(df):
    """"
    Objective :
    Function for preprocessing data, known missing, duplicated values and basic stastics every column in dataframe and duplicated row
    df is dataframe
    """
    try:
        import pandas as pd
        variables = pd.DataFrame(columns=['Variable','Number of unique values','Percent of Null(%)','Number of Duplicated','Type','Values'])
        for i, var in enumerate(df.columns):
            variables.loc[i] = [var, df[var].nunique(),df[var].isnull().sum()/df.shape[0]*100,df[var].duplicated(keep=False).sum(),df[var].dtypes,df[var].unique()]
        return (variables.set_index('Variable'))
    except:
        print('ada kesalahan penginputan')
```

## WOE and IV

```python
features = df_train.drop(['TARGET'],axis=1)
labels = df_train['TARGET']
```

```python
import pandas as pd
import numpy as np
from scipy.stats import spearmanr
from pandas.api.types import is_numeric_dtype
import warnings

warnings.filterwarnings("ignore", category=pd.errors.PerformanceWarning)

max_bin = 20
force_bin = 3
eps = 1e-10  # small number to avoid division by zero


def mono_bin(Y, X, n: int = max_bin):
    """
    Bin numeric variable X with respect to binary target Y and compute WOE/IV table.
    Y: pd.Series (binary: 0/1)
    X: pd.Series (numeric)
    """
    X = X.reset_index(drop=True)
    Y = Y.reset_index(drop=True)

    df = pd.DataFrame({"X": X, "Y": Y})
    justmiss = df[df["X"].isna()]
    notmiss = df[df["X"].notna()]

    # Try qcut with decreasing number of bins until monotonic correlation is found
    current_n = n
    grouped = None
    while current_n > 1:
        try:
            d1 = pd.DataFrame({
                "X": notmiss["X"],
                "Y": notmiss["Y"],
                "Bucket": pd.qcut(notmiss["X"], current_n, duplicates="drop")
            })
            grouped = d1.groupby("Bucket", as_index=True)
            means = grouped.mean()
            if len(means) < 2:
                current_n -= 1
                continue
            r, p = spearmanr(means["X"].values, means["Y"].values)
            if np.isnan(r):
                current_n -= 1
                continue
            # If correlation magnitude is close to 1, stop (monotonic)
            if np.abs(r) >= 0.9:  # threshold; original used abs(r) < 1 loop — use <=0.9 here
                break
            # if not monotonic, try fewer bins
            current_n -= 1
        except Exception:
            current_n -= 1

    # If grouping failed or produced only one bucket, force fallback bins
    if (grouped is None) or (len(grouped) <= 1):
        current_n = force_bin
        # use np.quantile to get bin edges
        bins = np.unique(np.quantile(notmiss["X"], np.linspace(0, 1, current_n)))
        if bins.size < 2:
            # create simple bins around min/max if quantile fails
            xmin, xmax = notmiss["X"].min(), notmiss["X"].max()
            if pd.isna(xmin) or pd.isna(xmax):
                bins = np.array([0.0, 1.0])
            else:
                mid = xmin + (xmax - xmin) / 2
                bins = np.array([xmin, mid, xmax])
        d1 = pd.DataFrame({
            "X": notmiss["X"],
            "Y": notmiss["Y"],
            "Bucket": pd.cut(notmiss["X"], bins=bins, include_lowest=True)
        })
        grouped = d1.groupby("Bucket", as_index=True)

    # Build summary table
    d3 = pd.DataFrame()
    d3["MIN_VALUE"] = grouped["X"].min()
    d3["MAX_VALUE"] = grouped["X"].max()
    d3["COUNT"] = grouped["Y"].count()
    d3["EVENT"] = grouped["Y"].sum()
    d3["NONEVENT"] = d3["COUNT"] - d3["EVENT"]
    d3 = d3.reset_index(drop=True)

    # append missing bucket if exists
    if len(justmiss) > 0:
        miss_row = {
            "MIN_VALUE": np.nan,
            "MAX_VALUE": np.nan,
            "COUNT": justmiss["Y"].count(),
            "EVENT": justmiss["Y"].sum()
        }
        miss_row["NONEVENT"] = miss_row["COUNT"] - miss_row["EVENT"]
        d3 = pd.concat([d3, pd.DataFrame([miss_row])], ignore_index=True)

    # safety for counts
    d3["COUNT"] = d3["COUNT"].astype(float).replace(0, eps)
    d3["EVENT"] = d3["EVENT"].astype(float)
    d3["NONEVENT"] = d3["NONEVENT"].astype(float)

    # distributions
    total_event = d3["EVENT"].sum()
    total_nonevent = d3["NONEVENT"].sum()
    total_event = total_event if total_event > 0 else eps
    total_nonevent = total_nonevent if total_nonevent > 0 else eps

    d3["EVENT_RATE"] = d3["EVENT"] / d3["COUNT"]
    d3["NON_EVENT_RATE"] = d3["NONEVENT"] / d3["COUNT"]
    d3["DIST_EVENT"] = d3["EVENT"] / total_event
    d3["DIST_NON_EVENT"] = d3["NONEVENT"] / total_nonevent

    # replace zeros before log
    d3["DIST_EVENT"] = d3["DIST_EVENT"].replace(0, eps)
    d3["DIST_NON_EVENT"] = d3["DIST_NON_EVENT"].replace(0, eps)

    d3["WOE"] = np.log(d3["DIST_EVENT"] / d3["DIST_NON_EVENT"])
    d3["IV"] = (d3["DIST_EVENT"] - d3["DIST_NON_EVENT"]) * d3["WOE"]

    d3["VAR_NAME"] = "VAR"
    cols = ["VAR_NAME", "MIN_VALUE", "MAX_VALUE", "COUNT", "EVENT", "EVENT_RATE",
            "NONEVENT", "NON_EVENT_RATE", "DIST_EVENT", "DIST_NON_EVENT", "WOE", "IV"]
    d3 = d3[cols].replace([np.inf, -np.inf], 0)
    # Aggregate IV to a single summary value (same as original)
    d3.loc[:, "IV"] = d3["IV"].sum()
    return d3


def char_bin(Y, X):
    """
    WOE/IV for categorical variable X
    """
    X = X.reset_index(drop=True)
    Y = Y.reset_index(drop=True)
    df = pd.DataFrame({"X": X, "Y": Y})
    justmiss = df[df["X"].isna()]
    notmiss = df[df["X"].notna()]

    grouped = notmiss.groupby("X", as_index=True)

    d3 = pd.DataFrame()
    d3["COUNT"] = grouped["Y"].count()
    # For MIN_VALUE / MAX_VALUE we store category value
    d3["MIN_VALUE"] = d3.index.astype(object)
    d3["MAX_VALUE"] = d3["MIN_VALUE"]
    d3["EVENT"] = grouped["Y"].sum()
    d3["NONEVENT"] = d3["COUNT"] - d3["EVENT"]
    d3 = d3.reset_index(drop=True)

    if len(justmiss) > 0:
        miss_row = {
            "MIN_VALUE": np.nan,
            "MAX_VALUE": np.nan,
            "COUNT": justmiss["Y"].count(),
            "EVENT": justmiss["Y"].sum()
        }
        miss_row["NONEVENT"] = miss_row["COUNT"] - miss_row["EVENT"]
        d3 = pd.concat([d3, pd.DataFrame([miss_row])], ignore_index=True)

    d3["COUNT"] = d3["COUNT"].astype(float).replace(0, eps)
    d3["EVENT"] = d3["EVENT"].astype(float)
    d3["NONEVENT"] = d3["NONEVENT"].astype(float)

    total_event = d3["EVENT"].sum()
    total_nonevent = d3["NONEVENT"].sum()
    total_event = total_event if total_event > 0 else eps
    total_nonevent = total_nonevent if total_nonevent > 0 else eps

    d3["EVENT_RATE"] = d3["EVENT"] / d3["COUNT"]
    d3["NON_EVENT_RATE"] = d3["NONEVENT"] / d3["COUNT"]
    d3["DIST_EVENT"] = d3["EVENT"] / total_event
    d3["DIST_NON_EVENT"] = d3["NONEVENT"] / total_nonevent

    d3["DIST_EVENT"] = d3["DIST_EVENT"].replace(0, eps)
    d3["DIST_NON_EVENT"] = d3["DIST_NON_EVENT"].replace(0, eps)

    d3["WOE"] = np.log(d3["DIST_EVENT"] / d3["DIST_NON_EVENT"])
    d3["IV"] = (d3["DIST_EVENT"] - d3["DIST_NON_EVENT"]) * d3["WOE"]

    d3["VAR_NAME"] = "VAR"
    cols = ["VAR_NAME", "MIN_VALUE", "MAX_VALUE", "COUNT", "EVENT", "EVENT_RATE",
            "NONEVENT", "NON_EVENT_RATE", "DIST_EVENT", "DIST_NON_EVENT", "WOE", "IV"]
    d3 = d3[cols].replace([np.inf, -np.inf], 0)
    d3.loc[:, "IV"] = d3["IV"].sum()
    return d3.reset_index(drop=True)


def data_vars(df: pd.DataFrame, target):
    """
    Compute WOE/IV for all variables in df except target.
    target: either column name (str) or pd.Series
    Returns: iv_df (detailed rows) and iv (per-variable summary)
    """
    # prepare target series
    if isinstance(target, str):
        y = df[target].astype(float)
        exclude = {target}
    elif isinstance(target, pd.Series):
        y = target.astype(float)
        exclude = set()
    else:
        raise ValueError("target must be column name or pd.Series")

    iv_list = []
    for col in df.columns:
        if col in exclude:
            continue
        x = df[col]
        # choose numeric or categorical path
        if is_numeric_dtype(x) and x.nunique(dropna=True) > 2:
            conv = mono_bin(y, x)
        else:
            conv = char_bin(y, x)
        conv = conv.copy()
        conv["VAR_NAME"] = col
        iv_list.append(conv)

    if len(iv_list) == 0:
        return pd.DataFrame(), pd.DataFrame(columns=["VAR_NAME", "IV"])

    iv_df = pd.concat(iv_list, ignore_index=True)
    iv = pd.DataFrame({"IV": iv_df.groupby("VAR_NAME")["IV"].max()}).reset_index()
    return iv_df, iv
```

```python
for var in transform_vars_list:
    small_df = final_iv[final_iv['VAR_NAME'] == var]
    transform_dict = dict(zip(small_df.MAX_VALUE,small_df.WOE))
    replace_cmd = ''
    replace_cmd1 = ''
    for i in sorted(transform_dict.items()):
        replace_cmd = replace_cmd + str(i[1]) + str(' if x <= ') + str(i[0]) + ' else '
        replace_cmd1 = replace_cmd1 + str(i[1]) + str(' if x == "') + str(i[0]) + '" else '
    replace_cmd = replace_cmd + '0'
    replace_cmd1 = replace_cmd1 + '0'
    if replace_cmd != '0':
        try:
            df[transform_prefix + var] = df[var].apply(lambda x: eval(replace_cmd))
        except:
            df[transform_prefix + var] = df[var].apply(lambda x: eval(replace_cmd1))
```

```python
from collections import OrderedDict
import pandas as pd
import sys
%matplotlib inline
    
def plot_pandas_style(styler):
    from IPython.core.display import HTML
    html = '\n'.join([line.lstrip() for line in styler.render().split('\n')])
    return HTML(html)

def highlight_max(s,color='yellow'):
    '''
    highlight the maximum in a Series yellow.
    '''
    is_max = s == s.max()
    return ['background-color: {}'.format(color) if v else '' for v in is_max]

def decile_labels(agg1,label,color='skyblue'):
    agg_dummy = pd.DataFrame(OrderedDict((('TOTAL',0),('TARGET',0),('NONTARGET',0),('PCT_TAR',0),('CUM_TAR',0),('CUM_NONTAR',0),('DIST_TAR',0),('DIST_NONTAR',0),('SPREAD',0))),index=[0])
    agg1 = agg1.append(agg_dummy).sort_index()
    agg1.index.name = label
    agg1 = agg1.style.apply(highlight_max, color = 'yellow', subset=['SPREAD'])
    agg1.bar(subset=['TARGET'], color='{}'.format(color))
    agg1.bar(subset=['TOTAL'], color='{}'.format(color))
    agg1.bar(subset=['PCT_TAR'], color='{}'.format(color))
    return(agg1)

def deciling(data,decile_by,target,nontarget):
    inputs = list(decile_by)
    inputs.extend((target,nontarget))
    decile = data[inputs]
    grouped = decile.groupby(decile_by)
    agg1 = pd.DataFrame({},index=[])
    agg1['TOTAL'] = grouped.sum()[nontarget] + grouped.sum()[target]
    agg1['TARGET'] = grouped.sum()[target]
    agg1['NONTARGET'] = grouped.sum()[nontarget]
    agg1['PCT_TAR'] = grouped.mean()[target]*100
    agg1['CUM_TAR'] = grouped.sum()[target].cumsum()
    agg1['CUM_NONTAR'] = grouped.sum()[nontarget].cumsum()
    agg1['DIST_TAR'] = agg1['CUM_TAR']/agg1['TARGET'].sum()*100
    agg1['DIST_NONTAR'] = agg1['CUM_NONTAR']/agg1['NONTARGET'].sum()*100
    agg1['SPREAD'] = (agg1['DIST_TAR'] - agg1['DIST_NONTAR'])
    agg1 = decile_labels(agg1,'DECILE',color='skyblue')
    return(plot_pandas_style(agg1))
```

```python
def scoring(features,clf,target):
    score = pd.DataFrame(clf.predict_proba(features)[:,1], columns = ['SCORE'])
    score['DECILE'] = pd.qcut(score['SCORE'].rank(method = 'first'),10,labels=range(10,0,-1))
    score['DECILE'] = score['DECILE'].astype(float)
    score['TARGET'] = target
    score['NONTARGET'] = 1 - target
    return(score)
```

```python
for var in transform_vars_list:
    small_df = final_iv[final_iv['VAR_NAME'] == var]
    transform_dict = dict(zip(small_df.MAX_VALUE,small_df.WOE))
    replace_cmd = ''
    replace_cmd1 = ''
    for i in sorted(transform_dict.items()):
        replace_cmd = replace_cmd + str(i[1]) + str(' if x <= ') + str(i[0]) + ' else '
        replace_cmd1 = replace_cmd1 + str(i[1]) + str(' if x == "') + str(i[0]) + '" else '
    replace_cmd = replace_cmd + '0'
    replace_cmd1 = replace_cmd1 + '0'
    if replace_cmd != '0':
        try:
            df_train[transform_prefix + var] = df_train[var].apply(lambda x: eval(replace_cmd))
            df_test[transform_prefix + var] = df_test[var].apply(lambda x: eval(replace_cmd))
        except:
            df_train[transform_prefix + var] = df_train[var].apply(lambda x: eval(replace_cmd1))
            df_test[transform_prefix + var] = df_test[var].apply(lambda x: eval(replace_cmd1))
```
