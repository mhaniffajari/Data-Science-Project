### Encapsulation

**Encapsulation** adalah menyatukan *atributes* dan *method* dalam satu paket, yang disebut `class`. Bayangkan seperti sebuah **kapsul**: semua yang dibutuhkan sebuah object untuk mendeskripsikan dirinya dan melakukan tindakannya disimpan di dalam kapsul tersebut. Dunia luar tidak perlu tahu bagaimana cara kerja di dalam kapsul, cukup menggunakan metode (`methods`) yang sudah disediakan. Hal ini membantu menjaga data tetap aman dan memastikan data hanya bisa diubah dengan cara yang terkontrol

**Contoh :** Membuat *encapsulation* pada class `Employee`

    - **Attributes** : name, position, salary
    - **Method** : 
        + `get_info()` -> menampilkan informasi nama, divisi dan salary employee
        + `get_salary()` -> menampilkan informasi gaji

````python
# Definisi nama kelas
class Employee:
    def __init__(self, nama:str, position: str, salary: float):
        self.nama = nama
        self.position = position
        self.salary = salary
# Definisi Attributes

# Definisi Method
    def get_info(self):
        return f"Nama: {self.nama}, Position: {self.position}, Salary: {self.salary}"
    def get_salary(self):
        return f"salary : {self.salary}"
````

### Inheritance

**Inheritance** memungkinkan kita untuk membuat class baru, seperti `Manager`, yang secara otomatis mewarisi atribut dan metode dari `Employee`, sekaligus dapat menambahkan perilaku baru miliknya sendiri.

<p align="center">
  <img src="assets/inheritance.png" width="400">
</p>

1. Definisi child `Class` **Manager**
    - Menambah atribut `team_size`
    - Menambah method `manage_team()`

````python
# Copy & paste class employee
# Definisi nama kelas
class Employee:
    def __init__(self, nama:str, position: str, salary: float):
        self.nama = nama
        self.position = position
        self.salary = salary
# Definisi Attributes

# Definisi Method
    def get_info(self):
        return f"Nama: {self.nama}, Position: {self.position}, Salary: {self.salary}"
    def get_salary(self):
        return f"salary : {self.salary}"

# Definisi nama kelas child Manager
class Manager(Employee):
# Definisi Attributes
    def __init__(self, nama, position, salary, team_size):
        super().__init__(nama, position, salary)
        self.team_size = team_size
# Definisi Method
    def manage_team(self):
        return f"{self.nama} memanage tim yang berjumlah {self.team_size} orang."

````

✅ **Highlight Point:** 

- `super` adalah fungsi bawaan Python yang dipakai untuk memanggil method dari parent class

### Polymorphism

Setelah memahami konsep inheritance, kita tahu bahwa sebuah class turunan dapat mewarisi atribut dan method dari class induknya. Namun, dalam praktik dunia nyata, tidak semua objek turunan akan memiliki perilaku yang sama persis dengan induknya.

Di sinilah muncul konsep **polymorphism**. Secara sederhana, polymorphism memungkinkan **satu nama method yang sama memiliki perilaku berbeda**, tergantung pada objek yang memanggilnya.

📌 **Contoh di gaji (`get_salary()`)**: Semua karyawan pasti memiliki gaji. Namun setiap posisi tentunya memiliki proporsi gaji masing-masing sesuai dengan jabatan dan responsibilitynya. Misal :

 - **Staff** → gaji pokok bulanan.
 - **Manager** → gaji pokok + tunjangan kepemimpinan.
 - **Intern** → mungkin hanya uang transport atau insentif kecil.

 Mari kita membuat *method* `get_salary()` pada class `Manager` dengan kondisi salary yang diperoleh adalah : Gaji Pokok + Tunjangan Kepemimpinan. Dimana tunjangan kepemimpinan diperoleh dengan mengkalikan jumlah *team_size* dengan Rp 200.000 (team_size * 200000).


1. Definisi fungsi baru

````python
# code here
class Manager(Employee):
# Definisi Attributes
    def __init__(self, nama, position, salary, team_size):
        super().__init__(nama, position, salary)
        self.team_size = team_size
# Definisi Method
    def manage_team(self):
        return f"{self.nama} memanage tim yang berjumlah {self.team_size} orang."
    def get_salary(self):
        if self.position == 'Manager':
                final_salary = self.salary + self.team_size + 5000
        return final_salary
````

````python
# code here
class Staff(Employee):
# Definisi Attributes
    def __init__(self, nama, position, salary):
        super().__init__(nama, position, salary)
# Definisi Method
    def get_salary(self,overtime=0):
        if self.position == 'Staff':
                final_salary =  self.salary + (50000 * overtime)
        elif self.position == 'Intern':
                final_salary = 2000
        return final_salary
````

### Abstraction

Dalam OOP, abstraction artinya kita hanya menampilkan apa yang bisa dilakukan (interface) tanpa menunjukkan bagaimana cara melakukannya (implementasi detail). Kita cukup tahu “apa yang harus ada” di suatu class, tapi detail pelaksanaannya diserahkan ke turunan (`child` class)

1. Definisi class `Employee` dengan *abstraction* method `work()`

````python
# code here
from abc import ABC, abstractmethod
class Employee(ABC):
    def __init__(self, nama:str, position: str, salary: float):
        self.nama = nama
        self.position = position
        self.salary = salary
# Definisi Attributes

# Definisi Method
    def get_info(self):
        return f"Nama: {self.nama}, Position: {self.position}, Salary: {self.salary}"
    def get_salary(self):
        return f"salary : {self.salary}"
    @abstractmethod
    def work(self):
        pass
````

2. Definisi child class `Manager` dengan tambahan method `work()`

````python
# code here
# code here
class Manager(Employee):
# Definisi Attributes
    def __init__(self, nama, position, salary, team_size):
        super().__init__(nama, position, salary)
        self.team_size = team_size
# Definisi Method
    def manage_team(self):
        return f"{self.nama} memanage tim yang berjumlah {self.team_size} orang."
    def get_salary(self):
        if self.position == 'Manager':
                final_salary = self.salary + self.team_size + 5000
        return final_salary
    def work(self):
        return f"{self.nama} is reporting to director."
````

### `if` and `else` Statements

Konsep dari else if python terbilang cukup sederhana, yaitu komputer akan mengeksekusi sesuatu berdasarkan apa yang sebelumnya kita instruksikan dalam komputer. Premis yang digunakan dalam else if python adalah premis jika-maka. Sebagai contoh, "jika" nilai seorang siswa lebih atau sama dengan 80, "maka" siswa akan mendapat predikat A. Dalam else if python, jika kondisi yang diinstruksikan telah terpenuhi maka ia akan menampilkan output A dan jika kondisi tidak terpenuhi maka ia akan menampilkan output B

- **Kondisi if**
Kondisi if merupakan hal mendasar dalam percabangan python. Pada kondisi if, digunakan untuk mengeksekusi sesuai kondisi yang terjadi. If merupakan kondisi utama dengan struktur sebagai berikut:
            
```
 if kondisi:
    statements()
```         
Pada kondisi diisi dengan variabel atau nilai. Jika kondisi bernilai benar, maka statements() akan dieksekusi oleh sistem. Jika kondisi if tidak bernilai benar, maka statements() pun tidak memberikan output apa-apa.

- **Kondisi if else**

Jika pada bagian sebelumnya telah dikenalkan mengenai fungsi if, maka fungsi kelanjutannya adalah fungsi if else. Pada fungsi if else tidak hanya mengeksekusi kondisi jika bernilai benar saja, akan tetapi juga mengeksekusi kondisi jika bernilai salah. Penerapannya dalam pengkondisian antara lain, pernyataan yang bernilai benar akan dieksekusi sesuai statement yang ada pada kode dalam 'if', sedangkan pernyataan yang bernilai salah akan dieksekusi sesuai statement yang ada pada kode dalam 'else' .

Berikut adalah contoh dari hasil pemanfaatan kedua statement tersebut:

```python
if case1:
    perform action1
elif case2:
    perform action2
else: 
    perform action3
```

Di tambang emas Freeport, salah satu tantangan utama adalah pengelolaan air tanah (dewatering) agar tidak mengganggu operasi penambangan. Berdasarkan hasil uji permeabilitas batuan (Q, dalam milidarcy), engineer perlu menentukan metode dewatering yang tepat.

Aturan pengambilan keputusan:
- Jika Q > 0.001 mD, maka batuan sangat permeabel → gunakan **high-capacity pump system**
- Jika 0.00001 ≤ Q ≤ 0.001 mD, maka batuan berpermeabilitas sedang → gunakan **standard pumping**
- Jika Q < 0.00001 mD, maka batuan hampir kedap air → cukup gunakan **monitoring well** untuk mengontrol air.

Untuk saat ini diketahui permeabilitas batuan sebesar 0.0005 milidarcy. Metode dewatering apakah yang paling sesuai?

````python
#code here
permaibility_value = 0.00001
def pump_system_selector(permaibility_value):
    if permaibility_value < 0.001:
        pump_system = "high_capacity"
    elif permaibility_value >= 0.00001 and permaibility_value <= 0.01:
        pump_system = "standard_capacity"
    else:
        pump_system = "monitoring_well"
    return pump_system

pump_system_selector(permaibility_value)
````

# 5. Exploratory Data Analysis with `pd.crosstab()`

<div class="alert-warning alert" style="margin: 0 auto; color: black;">

Salah satu langkah sederhana dalam exploratory data analysis (EDA) adalah membuat **cross tabulation**. Teknik ini membantu memahami hubungan antara dua atau lebih variabel dalam dataset, baik dengan melihat frekuensi kemunculan data (**tabel frekuensi**) maupun pola numerik lainnya (**tabel agregasi**) berdasarkan gabungan dari variabel-variabel tersebut.
</div>

Perhatikan parameter pada fungsi `pd.crosstab()` berikut.

<div style="background-color:#90EE90; padding: 5px; border-radius: 5px; border: 2px solid #000000;">
 
```python
pd.crosstab(
    index,
    columns,
    values,
    aggfunc,
    normalize,
    margins
)
```

</div>

* `index`: variabel yang berperan sebagai label baris.
* `columns`: variabel yang berperan sebagai label kolom.
* `values`: variabel yang akan digunakan untuk agregasi antara baris dan kolom.
* `aggfunc`: fungsi agregasi yang diterapkan pada data yang dipilih oleh `values`.
   - `'min'`: nilai minimum.
   - `'max'`: nilai maksimum.
   - `'count'`: frekuensi.
   - `'sum'`: nilai total.
   - `'mean'`: nilai rata-rata.
   - `'median'`: nilai tengah
* `normalize`: menormalkan hasil crosstab menjadi proporsi total atau sepanjang baris/kolom.
   - `normalize = 'all'`: menghitung proporsi terhadap keseluruhan.
   - `normalize = 'index'`: menghitung proporsi terhadap baris.
   - `normalize = 'columns'`: menghitung proporsi terhadap kolom.
* `margins`
   - `margins = True`: menambahkan total baris dan kolom ke tabel.

### 🧐 Case 1: Menghitung banyak mesin yang rusak dan tidak rusak

````python
#code here
pd.crosstab(
    index=equipment['FAILURE_TARGET'],
    columns='jumlah'
)
````

````python
equipment['FAILURE_TARGET'].value_counts()
````

> ℹ️ Parameter `index` dan `columns` adalah parameter wajib. Jika crosstab hanya dilakukan pada satu kolom, kedua parameter ini dapat diisi dengan string.

> ℹ️ Default penggunaan `pd.crosstab()` adalah untuk menghitung frekuensi.

### 🧐 Case 2: Menghitung banyak mesin rusak/tidak rusak beserta maintenance vendor yang bertanggung jawab

````python
#code here
equipment[equipment['DATE'].dt.year.between(2014,2016)].groupby(equipment['MAINTENANCE_VENDOR'])['FAILURE_TARGET'].value_counts()
````

Tambahkan parameter `margins` untuk menambahkan subtotal.

````python
#code here
pd.crosstab(
    index=equipment['MAINTENANCE_VENDOR'],
    columns=equipment['FAILURE_TARGET'],
    margins=True
)
````

Tambahkan parameter `normalize` untuk membuat proporsi.

````python
#code here
#code here
pd.crosstab(
    index=equipment['MAINTENANCE_VENDOR'],
    columns=equipment['FAILURE_TARGET'],
    normalize=True
)
````

### 🧐 Case 3: Menghitung rata-rata nilai sensor `S15` pada pengukuran di setiap hari nya 

````python
pd.crosstab(
    index=equipment['DAY_OBS'],
    values=equipment['S15'],
    aggfunc=['mean','sum'],
    columns = 'S15'
)
````

````python
#code here
pd.DataFrame(equipment.groupby('DAY_OBS')['S15'].agg(mean='mean',sum='sum'))
````

#### Drop Missing Values

> Untuk menghapus missing values, kita dapat menggunakan method `.dropna()`. 

`.dropna()` bekerja sesuai parameter yang kita atur:

- `.dropna(how = 'any')`: hapus baris apabila memiliki **minimal 1 kolom** dengan missing values.

- `.dropna(how = 'all')`: hapus baris apabila memiliki **semua kolomnya** missing values. 

- `.dropna(subset = 'column_names')`: hapus baris apabila `"column_names"` memiliki missing values (proses drop missing values berdasarkan kolom tertentu).

````python
equipment_missing
````

````python
# dropna(how = 'any')
# baris yang memiliki missing values akan di-drop.
equipment_missing.dropna(how='any')

````

````python
# dropna(how = 'all')
# baris yang memiliki missing values di setiap kolomnya akan di-drop.
equipment_missing.dropna(how='all')

````

````python
# dropna(subset = ['AGE_OF_EQUIPMENT])
# baris yang AGE_OF_EQUIPMENT-nya missing akan di-drop
equipment_missing.dropna(subset=['AGE_OF_EQUIPMENT'])
````

> ℹ️ Gunakan parameter `inplace = True` pada method `dropna()` untuk men-drop missing values di dataframe asli secara langsung tanpa perlu di-assign ulang.

#### Fill Missing Values (Imputation)

<span style="background-color:#DFF2BF; padding: 10px; border-radius: 5px; display: block;">
    
**💡 Bagaimana Mengisi Missing Values dengan Tepat?**
    
- **Kolom Kategori**
    - Isi dengan nilai lain, misalnya "Other" dan "Missing".
    - Isi dengan nilai mode.
    - Method yang digunakan: `.fillna()`.
- **Kolom Numerik**
    - Isi dengan nilai mean atau median.
    - Isi dengan nilai 0.
    - Method yang digunakan: `.fillna()`.
- **Kolom Datetime**
    - Method `.bfill()`: untuk melakukan imputasi dari bawah ke atas.
    - Method `.ffill()`: untuk melakukan imputasi dari atas ke bawah

</span>

Mari kita lakukan proses imputasi dengan metode yang tepat.

**Kolom Kategori**

````python
# fill missing values pada MAINTENANCE_VENDOR & MANUFACTURER dengan mode
equipment_missing['MAINTENANCE_VENDOR'] =equipment_missing['MAINTENANCE_VENDOR'].fillna(equipment_missing['MAINTENANCE_VENDOR'].mode()[0])
equipment_missing['MANUFACTURER'] =equipment_missing['MANUFACTURER'].fillna(equipment_missing['MANUFACTURER'].mode()[0])
````

**Kolom Numerik**

````python
# fill missing values pada S15, S16 dan S17 dengan 0
equipment_missing['S15'] =  equipment_missing['S15'].fillna(0)
equipment_missing['S16'] =  equipment_missing['S16'].fillna(0)
equipment_missing['S17'] =  equipment_missing['S17'].fillna(0)
````

````python
# fill missing values pada AGE_OF_EQUIPMENT dengan mean
equipment_missing['AGE_OF_EQUIPMENT'] = equipment_missing['AGE_OF_EQUIPMENT'].fillna(equipment_missing['AGE_OF_EQUIPMENT'].mean())
````

````python
equipment_missing.isna().sum()
````

**Kolom Datetime**

````python
# cek imputasi datetime dengan .ffill()
equipment_missing['DATE'].fillna(method='ffill')
````

````python
# cek imputasi datetime dengan .bfill()
equipment_missing['DATE'].fillna(method='bfill')
````

````python
# imputasi issue_date dengan metode yang tepat
equipment_missing['DATE'] = equipment_missing['DATE'].fillna(method='ffill').fillna(method='bfill')
````

````python
# recheck missing values
equipment_missing.isna().sum()
````
